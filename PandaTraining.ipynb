UC2:DATA CLEANING AND VALIDATION


Step 1: Check for Missing Values

```python
# Check missing values

print("===== MISSING VALUES: DELIVERIES =====")
print(deliveries_df.isnull().sum())

print("\n===== MISSING VALUES: MATCHES =====")
print(matches_df.isnull().sum())
```

Step 2: Handle / Report Inconsistencies

```python
# Create copies
deliveries_df_cleaned = deliveries_df.copy()
matches_df_cleaned = matches_df.copy()

# Handle missing values (correct way)

matches_df_cleaned['winner'] = matches_df_cleaned['winner'].fillna('Unknown')
matches_df_cleaned['player_of_match'] = matches_df_cleaned['player_of_match'].fillna('Unknown')
matches_df_cleaned['city'] = matches_df_cleaned['city'].fillna('Unknown')

print("Missing values handled for key columns")
```

Step 3: Validation (IDs + Data Types)

```python


deliveries_df_cleaned['match_id'] = deliveries_df_cleaned['match_id'].astype(int)
matches_df_cleaned['id'] = matches_df_cleaned['id'].astype(int)

print("Data types validated successfully")



delivery_ids = set(deliveries_df_cleaned['match_id'].unique())
match_ids = set(matches_df_cleaned['id'].unique())

missing_ids = delivery_ids - match_ids

if len(missing_ids) == 0:
    print("All match IDs are consistent between datasets")
else:
    print(f"Inconsistency found: {len(missing_ids)} match_ids missing in matches dataset")
```

Step 1: Create New Column (Total Runs per Ball)

```python


deliveries_df_cleaned['total_runs'] = (
    deliveries_df_cleaned['batsman_runs'] +
    deliveries_df_cleaned['extra_runs']
)

print("total_runs column created successfully")
deliveries_df_cleaned[['batsman_runs', 'extra_runs', 'total_runs']].head()
```

Step 2: Standardize Column Names

```python


deliveries_df_cleaned.columns = deliveries_df_cleaned.columns.str.lower().str.strip()
matches_df_cleaned.columns = matches_df_cleaned.columns.str.lower().str.strip()

print("Column names standardized")
print(deliveries_df_cleaned.columns)
print(matches_df_cleaned.columns)
```

Step 3: Merge Datasets

```python


ipl_df = deliveries_df_cleaned.merge(
    matches_df_cleaned,
    left_on="match_id",
    right_on="id",
    how="inner"
)

print("Datasets merged successfully")

# Final output
print("\nUnified DataFrame ready for analysis")
print("Shape:", ipl_df.shape)

ipl_df.head()
```

UC4:SCORE ANALYSIS

Step 1:Total Runs per Match

```python
total_runs_per_match = ipl_df.groupby('match_id')['total_runs'].sum().reset_index()
total_runs_per_match.columns = ['match_id', 'total_runs']

total_runs_per_match.head()
```

Step 2:Runs per Team per Match

```python
runs_per_team_match = ipl_df.groupby(['match_id', 'batting_team'])['total_runs'].sum().reset_index()

runs_per_team_match.head()
```

Step 3:Top 10 Batters

```python
top_batters = ipl_df.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(10)

top_batters
```

Step 4:Strike Rate of Batters

```python
batters_stats = ipl_df.groupby('batter').agg(
    runs=('batsman_runs', 'sum'),
    balls=('batsman_runs', 'count')
).reset_index()

batters_stats['strike_rate'] = (batters_stats['runs'] / batters_stats['balls']) * 100

batters_stats.sort_values(by='strike_rate', ascending=False).head(10)
```

Step 5:Top 10 bowlers by economy

```python
bowler_stats = ipl_df.groupby('bowler').agg(
    runs=('total_runs', 'sum'),
    balls=('bowler', 'count')
).reset_index()

bowler_stats['overs'] = bowler_stats['balls'] / 6
bowler_stats['economy'] = bowler_stats['runs'] / bowler_stats['overs']

bowler_stats.sort_values(by='economy').head(10)
```

Step 6:Most Consistent Batters

```python
consistency = ipl_df.groupby(['batter', 'match_id'])['batsman_runs'].sum().reset_index()

consistency_stats = consistency.groupby('batter').agg(
    avg_runs=('batsman_runs', 'mean'),
    matches=('match_id', 'count')
).reset_index()

consistent_batters = consistency_stats[
    consistency_stats['matches'] >= 10
].sort_values(by='avg_runs', ascending=False)

consistent_batters.head(10)
```

Step 7:Highest Individual score in a match

```python
highest_score = ipl_df.groupby(['match_id', 'batter'])['batsman_runs'].sum().reset_index()

highest_score.sort_values(by='batsman_runs', ascending=False).head(1)
```

Step 8:Boundary Analysis

```python
boundaries = ipl_df[ipl_df['batsman_runs'].isin([4, 6])]

boundary_counts = boundaries.groupby('batter')['batsman_runs'].count().sort_values(ascending=False)

boundary_counts.head(10)
```

Step 9:Boundary Percentage

```python
batter_total = ipl_df.groupby('batter')['batsman_runs'].sum()
batter_boundaries = ipl_df[ipl_df['batsman_runs'].isin([4, 6])].groupby('batter')['batsman_runs'].sum()

boundary_percent = (batter_boundaries / batter_total) * 100

boundary_percent.sort_values(ascending=False).head(10)
```

Step 10:Dot Ball Analysis

```python
dot_balls = ipl_df[ipl_df['batsman_runs'] == 0]

dot_count = dot_balls.groupby('bowler')['batsman_runs'].count().sort_values(ascending=False)

dot_count.head(10)
```

Step 11:Runs per Over Analysis

```python
runs_per_over = ipl_df.groupby('over')['total_runs'].mean().reset_index()

runs_per_over.columns = ['over', 'avg_runs']

runs_per_over
```

Step 12:Powerplay(1-6 overs)

```python
powerplay = ipl_df[ipl_df['over'] <= 6]

powerplay_runs = powerplay.groupby('batting_team')['total_runs'].sum().sort_values(ascending=False)

powerplay_runs
```

Step 13:Death overs

```python
death_overs = ipl_df[(ipl_df['over'] >= 16) & (ipl_df['over'] <= 20)]

death_runs = death_overs.groupby('batting_team')['total_runs'].sum().sort_values(ascending=False)

death_runs
```

Step 14:Run Distribution per Inning

```python
inning_runs = ipl_df.groupby('inning')['total_runs'].sum()

inning_runs
```

Step 15:Toss Impact Analysis

```python
toss_runs = ipl_df.groupby('toss_winner')['total_runs'].sum().sort_values(ascending=False)

toss_runs
```

Step 16:Player of the Match Contribution

```python
pom_runs = ipl_df.groupby(['match_id', 'player_of_match', 'batter'])['batsman_runs'].sum().reset_index()

pom_runs.head()
```

Step 17:Venue-wise Analysis

```python
venue_runs = ipl_df.groupby('venue')['total_runs'].mean().sort_values(ascending=False)

venue_runs
```

Step 18:City-wise Scoring Trends

```python
city_runs = ipl_df.groupby('city')['total_runs'].mean().sort_values(ascending=False)

city_runs
```

Step 19:Season-wise Run Trends

```python
season_runs = ipl_df.groupby('season')['total_runs'].sum().reset_index()

season_runs
```

Step 20:Winning Team Analysis

```python
match_scores = ipl_df.groupby(['match_id', 'batting_team'])['total_runs'].sum().reset_index()

winner_analysis = match_scores.loc[match_scores.groupby('match_id')['total_runs'].idxmax()]

winner_analysis.head()
```

Step 1:Most Consistent Batter

```python

batter_match_runs = ipl_df.groupby(['batter', 'match_id'])['batsman_runs'].sum().reset_index()


batter_consistency = batter_match_runs.groupby('batter').agg(
    avg_runs=('batsman_runs', 'mean'),
    matches=('match_id', 'count')
).reset_index()

consistent_batters = batter_consistency[batter_consistency['matches'] >= 10]


most_consistent_batter = consistent_batters.sort_values(by='avg_runs', ascending=False).head(1)

most_consistent_batter
```

Step 2:Best Death-Over Team & High-Scoring Venues

```python


death_overs = ipl_df[(ipl_df['over'] >= 16) & (ipl_df['over'] <= 20)]

death_team_runs = death_overs.groupby('batting_team')['total_runs'].sum().reset_index()

best_death_team = death_team_runs.sort_values(by='total_runs', ascending=False).head(1)




venue_runs = ipl_df.groupby('venue')['total_runs'].mean().reset_index()

high_scoring_venue = venue_runs.sort_values(by='total_runs', ascending=False).head(1)


print("Best Death-Over Team:")
display(best_death_team)

print("\nHigh Scoring Venue:")
display(high_scoring_venue)
```

Step 3: Text-Based Insights

```python


batter_name = most_consistent_batter.iloc[0]['batter']
avg_runs = most_consistent_batter.iloc[0]['avg_runs']

team_name = best_death_team.iloc[0]['batting_team']
death_runs = best_death_team.iloc[0]['total_runs']

venue_name = high_scoring_venue.iloc[0]['venue']
venue_avg = high_scoring_venue.iloc[0]['total_runs']


print(" INSIGHTS \n")

print(f"1. Most Consistent Batter: {batter_name} with an average of {avg_runs:.2f} runs per match.")

print(f"2. Best Death-Over Team: {team_name} scored the highest runs ({death_runs}) in overs 16–20.")

print(f"3. High-Scoring Venue: {venue_name} has the highest average runs ({venue_avg:.2f}) per match.")
```
